```python
# instance variable and method with self 

class Student:
    def __init__(self,name,age,grade):
        self.name = name
        self.age = age
        self.grade = grade
    def details(self):
        print(f" Student Name:{self.name }")
        print(f" Student Age:{self.age }")
        print(f" Student Grade:{self.grade }")
    def is_eligible(self):
        if self.age>=15:
            print(f"{self.name} is eligible")
        else:
            print(f"{self.name} is not eligible")
```

```python
# accessing Instance variables and Methods ,self is the object on which methods are call

student = Student("Ayesha",25,"B")
student.details()
student.is_eligible()

print(student.age) # access variable 
```

Getter and Setters

```python
class Student:
    def grade(self):
        return self.__grade
    def __init__(self,name,grade):
        self.__grade = grade
    def get_grade(self):
        return self.__grade
    def set_grade(self,grade):
        if  0<= grade <=100:
            self.__grade = grade
        else:
            print("invalid")
            
```
